## Introduction:
The final output convolution layers of a trained image classification network contain detailed semantic information about the input image. These feature maps, although low resolution (e.g. 7 x 7) also contain spatial information about the location of the primary object in the image. In the class activation mapping technique we wish to extract this spatial information pertaining to the identified class and overlay this activation on the input image as shown in the figure below.

```

imagenette_dict={0:'Tench Fish',217:'English Springer',482:'Casette Player',
                491:'Chain Saw',497:'Church',566:'French Horn',
                 569:'Grabage Truck',571:'Gas Pump',574:'Golfball',
                 662:'Modem', 701:'Parachute',864:'Tow Truck', 867:'Trailer Truck'}
mean = [0.485, 0.456, 0.406]  # Example mean used for normalization
std = [0.229, 0.224, 0.225]   # Example std used for normalization

def unnormalize_image(image, mean, std):
    """
    Unnormalize a PyTorch image tensor.

    Args:
        image (torch.Tensor): The normalized image tensor of shape [C, H, W].
        mean (list or tuple): The mean used for normalization.
        std (list or tuple): The standard deviation used for normalization.

    Returns:
        torch.Tensor: The unnormalized image tensor.
    """
    # Remove the batch dimension
    image = image.squeeze(0)

    # Check if the image is in the shape [C, H, W]
    if image.dim() == 3:
        # Unnormalize the image
        for t, m, s in zip(image, mean, std):
            t.mul_(s).add_(m)
    else:
        raise ValueError("Expected image tensor of shape [C, H, W]")

    # Convert from [C, x, y] to [x, y, C]
    image = image.permute(1, 2, 0)

    return image
```

# Class Activion Mapping
You can implement the class activation mapping algorithm in the code cell below.



```

import matplotlib.pyplot as plt
import numpy as np

# Initialize Figure
fig = plt.figure(figsize=(10,10), frameon=False)


# Iterate over the dataiter to produce a few examples

# Ensure model is eval mode to disable batch normalisation during inference.
model.eval()
for i in range(10):

    ######################################################
    # Your Code here #####################################
    ######################################################
    # 1. Get image batch
    img_batch, label_batch = next(dataiter)
    img_batch = img_batch.to(device)

    # 2. Forward pass
    out = model(img_batch)
    features = out['op_features']        # [1, 2048, 7, 7]
    logits = out['op_logits']            # [1, 1000]

    # 3. Predicted class
    curr_class = torch.argmax(logits, dim=1)

    # 4. Extract classifier weights
    fc_weights = model.state_dict()['fc.weight']   # [1000, 2048]
    # Get weights of the predicted class, by selecting the row for the predicted class (curr_class), we get the importance weights for that class.
    class_weights = fc_weights[curr_class]         # [1, 2048]

    # 5. Compute CAM
    features = features.squeeze(0)   # [2048, 7, 7]
    ## torch.matmul(class_weights, …) computes a weighted sum across channels:
    ### Each channel’s activation map is multiplied by its importance weight.
    ### Then all weighted maps are summed together.
    cam = torch.matmul(class_weights, features.reshape(2048, -1))  # [1, 49]
    ## 7×7 heatmap showing which spatial locations contributed most to the predicted class.
    cam = cam.reshape(7, 7)

    # Normalize CAM
    cam = cam - cam.min()
    cam = cam / cam.max()

    # 6. Upsample CAM to image size
    act_map_us = F.interpolate(cam.unsqueeze(0).unsqueeze(0),
                               size=(224,224),
                               mode='bilinear',
                               align_corners=False).squeeze()


    ######################################################
    # END Your Code here #####################################
    ######################################################


    # N. Plotting
    #img =
    # N. Plotting
    def get_value(dictionary, key):
        try:
            return dictionary[key]
        except KeyError:
            return "Error"

    label = get_value(imagenette_dict, int(curr_class.cpu().numpy()))

    ax = fig.add_subplot(4, 4, i + 1, xticks=[], yticks=[])
    ax.title.set_text(label)

    # Use the current batch image
    img = unnormalize_image(img_batch.cpu(), mean, std)

    ax.imshow(img)
    ax.imshow(act_map_us.detach().cpu().numpy(), cmap=plt.cm.jet, alpha=0.4)

plt.show()



```

**My approach was in summary was:**


*   Pass the image through ResNet‑50.

*   Extract the last convolutional feature maps, which contain spatial and semantic information.

*   Use the classifier weights to identify which channels matter for the predicted class.

*   Compute a weighted sum of the feature maps, producing a coarse CAM (7×7).

*   Upsample the CAM to match the input image size (224×224).

*   Overlay the CAM heatmap on the original image for visualization.


**In a little bit more details:**

**Forward pass through the model**

- features: the output of the last convolutional layer (2048 channels, spatial size 7×7).

- logits: the raw classifier scores for 1000 ImageNet classes.


**PRedicted class**
  
- Finds the index of the class with the highest score.

- This is the network’s predicted label for the image.


**Extract classifier weights**

- The final fully connected (FC) layer has one row of weights per class.

- Each row tells the network how much each of the 2048 feature channels contributes to that class.


**Compute CAM**

- Remove the batch dimension → feature maps are [2048, 7, 7].

- Flatten spatial dimensions → [2048, 49].

- Multiply by the class weights → weighted sum across channels.

- Result: a 7×7 heatmap showing which spatial locations contributed most to the predicted class.


**Normalize CAM**

- Scale values to the range [0,1].

- Ensures consistent visualization (bright = strong activation, dark = weak).


**Upsample CAM to image size**

- unsqueeze adds batch and channel dimensions so PyTorch can interpolate.

- F.interpolate resizes CAM to 224×224 (same as input image).

- .squeeze() removes extra dimensions → final CAM is [224, 224].




We can check that CAM highlights where in the image the network "looked" to mkae its decision. For example: "French Horn". the CAM will light up the horn image in the image. This bridges classification (what the model predicts) with localization (where the evidence is).

**Spatial info:** preserved in *features = out['op_features'] (7×7 grid)*.

**Class importance:** extracted in *class_weights = fc_weights[curr_class]*.

**Interpretability:** achieved in *cam = torch.matmul(class_weights, features.reshape(2048, -1))*, where spatial maps are weighted by class importance.
